```python
import numpy as np 
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
```

```python
df = pd.DataFrame({
    "study_hours": [1, 2, 3, 4, 5, 6, 7, 8, 2, 9],
    "attendance": [40, 50, 60, 65, 75, 80, 85, 95, 55, 98],
    "assignment_score": [30, 40, 50, 55, 70, 75, 80, 95, 45, 98],
    "result": [0, 0, 0, 1, 1, 1, 2, 2, 0, 2]
})
```

```python
df.head()
```

```python
df.columns
```

```python
x=df[['study_hours', 'attendance', 'assignment_score']]
y=df[['result']]
```

```python
xmin=x.min()
xmax=x.max()
x_scaled=(x-xmin)/(xmax - xmin + 1e-8)
```

```python
x
```

```python
x_scaled
```

```python
x_train,x_text,y_train,y_test=train_test_split(x_scaled,y,random_state=42,test_size=0.25)
```

```python
x_train
```

```python
x_train.shape
```

```python
#Simple Deep learnign model
model=keras.Sequential([
    layers.Input(shape=(x_train.shape[1],)),
    layers.Dense(8,activation='relu'),
    layers.Dense(3,activation='sigmoid')
])
```

```python
model.compile(optimizer='opt',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
history_VGD=model.fit(x_train,y_train,epochs=100,batch_size=len(x_train),verbose=1)

history_SGD=model.fit(x_train,y_train,epochs=100,batch_size=1,verbose=1)

history_MGD=model.fit(
    x_train.values,y_train.values,
    validation_data=(x_text.values,y_test.values),
    epochs=100,batch_size=100,verbose=1

)
```
